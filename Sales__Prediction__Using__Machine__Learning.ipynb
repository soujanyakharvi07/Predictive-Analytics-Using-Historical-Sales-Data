step 5-Prepare Data for Machine Learning
Convert Product Category into numbers

```python
df = pd.get_dummies(
    df,
    columns=["Product_Category"],
    drop_first=True
)

df.head()
```

Select input columns

```python
X = df.drop(columns=["Date", "Sales"])
```

Select target column

```python
y = df["Sales"]
```

Check X and y

```python
print("Input shape:", X.shape)
print("Target shape:", y.shape)
```

Split the data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

Step 6: Build and Train the Prediction Model

Import Linear Regression

```python
from sklearn.linear_model import LinearRegression
```

create the model

```python
model = LinearRegression()
```

train the model

```python
model.fit(X_train, y_train)
```

make predictions

```python
y_pred = model.predict(X_test)
```

Display some predictions

```python
comparison = pd.DataFrame({
    "Actual Sales": y_test.values,
    "Predicted Sales": y_pred
})

comparison.head(10)
```

Step 7: Evaluate the Model Accuracy

Import evaluation metrics

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

Calculate the metrics

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)
```

Display the results

```python
print("Model Evaluation Results")
print("------------------------")
print("MAE  :", mae)
print("MSE  :", mse)
print("RMSE :", rmse)
print("R² Score :", r2)
```

Step 9: Predict Future Sales

Create future data

```python
future_data = pd.DataFrame({
    "Units_Sold": [60, 70, 80, 90, 100],
    "Unit_Price": [1500, 1600, 1700, 1800, 1900],
    "Marketing_Spend": [2000, 2200, 2400, 2600, 2800],
    "Customers": [250, 280, 300, 330, 350],
    "Discount_Percent": [10, 8, 7, 5, 5]
})
```

```python
for column in X.columns:
    if column not in future_data.columns:
        future_data[column] = 0

future_data = future_data[X.columns]
```

```python
future_predictions = model.predict(future_data)

future_predictions
```

Display the results

```python
results = pd.DataFrame({
    "Future_Record": range(1, 6),
    "Predicted_Sales": future_predictions
})

results["Predicted_Sales"] = results["Predicted_Sales"].round(2)

results
```
